### Caricamento del task

Carica un task di `run/llm_tasks` con lo stesso meccanismo di `run_grpo.py`: import del modulo `llm_tasks.<TASK_NAME>` + chiamata a `make_dataset(n_samples, seed)`. Per cambiare dataset basta cambiare `TASK_NAME`.

```python
import os, sys, importlib

os.environ.setdefault("HF_HOME", "/storage/fis1/hf_cache")
os.environ.setdefault("HF_HUB_OFFLINE", "1")
sys.path.insert(0, os.path.abspath("../run"))

TASK_NAME = "simple_sums"   # nome modulo in run/llm_tasks: simple_sums | gsm8k | multiarith | ...

task = importlib.import_module(f"llm_tasks.{TASK_NAME}")
train, test = task.make_dataset(200, seed=42)
print(f"train={len(train)}  test={len(test)}  colonne={train.column_names}")
```

### Print grezzo

Ogni elemento del dataset è un dizionario:
- `prompt`: conversazione in formato chat, lista di messaggi `{'role': ..., 'content': ...}` (system + user, col system che chiede la risposta in `\boxed{}`)
- `solution`: risposta corretta come stringa (es. `'41'`), letta da `accuracy_reward`

Tutti i task della repo espongono le stesse colonne `prompt`/`solution`.

```python
for row in train.select(range(5)):
    print(row, "\n")
```

### Print leggibile

Una riga per messaggio del prompt (`role: content`), poi le colonne extra (`solution` ecc.); riga vuota tra un campione e l'altro.

```python
for row in train.select(range(10)):
    for m in row["prompt"]:
        print(f"{m['role']}: {m['content']}")
    for k, v in row.items():
        if k != "prompt":
            print(f"{k}: {v}")
    print()
```
